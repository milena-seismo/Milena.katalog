```python
# Find the most expensive set in the "Creator Expert" theme.
Logika: FILTRIRAJ → SORTIRAJ → PRIKAŽI PRVI

lego[lego["Theme"] == "Creator Expert"].sort_values(
    by="Current_Price",
    ascending=False
).head(1)

Objašnjenje:

•	lego["Theme"] == "Creator Expert" - pronalazi setove određene teme.
•	sort_values() - sortira setove po cijeni.
•	by="Current_Price" - koristi trenutnu cijenu.
•	ascending=False - od najskupljeg ka najjeftinijem.
•	head(1) - prikazuje samo prvi red.
Važno:
•	ascending=True - od najmanjeg ka najvećem.
•	ascending=False - od najvećeg ka najmanjem.

Za ovaj zadatak ne treba groupby(), jer tražimo jedan konkretan set, a ne statistiku grupe.
```

```python
# Find the average price AND the maximum number of pieces for each theme group in a single command.
Logika: GRUPIŠI → IZRAČUNAJ DVIJE STATISTIKE
lego.groupby("Theme_Group").agg(
    average_price=("USD_MSRP", "mean"),
    max_pieces=("Pieces", "max")
)
#Objašnjenje:
•	groupby("Theme_Group") - formira LEGO grupe.
•	agg() - omogućava više statističkih operacija.
•	average_price - naziv nove kolone u rezultatu.
•	("USD_MSRP", "mean") - prosječna preporučena cijena.
•	max_pieces - naziv druge nove kolone.
•	("Pieces", "max") - najveći broj djelova.

# Ako želimo prosječan broj djelova umjesto maksimalnog:
lego.groupby("Theme_Group").agg(
    average_price=("USD_MSRP", "mean"),
    average_pieces=("Pieces", "mean")
)
```
